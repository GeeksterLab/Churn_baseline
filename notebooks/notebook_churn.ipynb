## PREPROCESSING

### Feature Engineering

#### Encoding categoricals features

```python
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, RobustScaler
```

```python
# 0 = absence de l’événement
# 1 = présence de l’événement

df["Churn_encoded"] = df["Churn"].map({
    "No": 0,
    "Yes": 1,
})
```

```python
X = df.drop(columns=["Churn", "Churn_encoded", "customerID", "TotalCharges"])
y = df["Churn_encoded"]
```

```python
# most_frequent → most frequent category → classic & simple
# contant + fill_value="" → replace by "missing" → better if the info missing is importante
# mean → numeric
# median → numeric

cat_transformer = make_pipeline(
    SimpleImputer(strategy= "most_frequent"),
    OneHotEncoder(sparse_output=False, drop='if_binary', handle_unknown="ignore")
)

cat_cols = X.select_dtypes(exclude=["int64", "float64"]).columns.tolist()
```

#### Encoding numerical features

```python
num_transformer = make_pipeline(
    SimpleImputer(strategy="median"),
    RobustScaler(),
)

num_cols= X.select_dtypes(include=["int64", "float64"]).columns.tolist()
```

```python
from sklearn.compose import make_column_transformer
from sklearn.linear_model import LogisticRegression
```

```python
preprocessing = make_column_transformer(
    (cat_transformer, cat_cols),
    (num_transformer, num_cols),
).set_output(transform="pandas")
```

```python
model = LogisticRegression(
    solver='lbfgs',
    C=1.0,
    max_iter=200,
)
```

```python
pipe_supervised = make_pipeline(
    preprocessing,
    model,
)
```

## TRAINING

### Train Test Split

```python
from sklearn.model_selection import train_test_split, cross_validate, RandomizedSearchCV
from scipy.stats import loguniform
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
)

pipe_supervised.fit(X_train, y_train)
```

### Cross Validate

```python
score_baseline = cross_validate(
    pipe_supervised,
    X_train,
    y_train,
    cv=5,
    scoring=[
        "accuracy",
        "recall",
        "precision",
    ]

)
```

### Best Params

```python
param_grid= {
    "logisticregression__C": loguniform(1e-3, 1e3),
    "logisticregression__class_weight":[None, "balanced"],
    "logisticregression__solver": ["lbfgs","liblinear", "saga"],
    "logisticregression__max_iter": [500, 1000, 2000, 3000],

}

search = RandomizedSearchCV(
    pipe_supervised,
    param_distributions=param_grid,
    cv=5,
    scoring=[
        "accuracy",
        "precision",
        "recall",
        "f1",
        "roc_auc",
    ],
    refit="recall",
    n_jobs=-1,
    error_score="raise"
)

search.fit(X_train,y_train)
```

```python
display(search.best_estimator_, search.best_params_, search.best_score_)
```

```python
results = pd.DataFrame(search.cv_results_)

results[
    [
        "mean_test_accuracy",
        "mean_test_precision",
        "mean_test_recall",
        "mean_test_f1",
        "mean_test_roc_auc",
        "params",
    ]
].sort_values("mean_test_recall", ascending=False).head()
```

## CHECK RESULTS

### Diagnostic

```python
# baseline
pipe_supervised.fit(X_train, y_train)
y_pred_baseline = pipe_supervised.predict(X_test)

# new model
search.fit(X_train, y_train)
y_pred_tuned = search.predict(X_test)

y_proba_baseline = pipe_supervised.predict_proba(X_test)[:,1]
y_proba_tuned = search.best_estimator_.predict_proba(X_test)[:,1]
```

### Comparer les scores

```python
from sklearn.metrics import recall_score, precision_score, accuracy_score, f1_score, roc_auc_score
```

```python
score_compares = pd.DataFrame({
    "model": ["baseline", "tuned"],
    "recall": [
        recall_score(y_test, y_pred_baseline),
        recall_score(y_test, y_pred_tuned),
    ],
    "precision": [
        precision_score(y_test, y_pred_baseline),
        precision_score(y_test, y_pred_tuned),
    ],
    "accuracy": [
        accuracy_score(y_test, y_pred_baseline),
        accuracy_score(y_test, y_pred_tuned),
    ],
    "f1": [
        f1_score(y_test, y_pred_baseline),
        f1_score(y_test, y_pred_tuned),
    ],
    "roc-auc": [
        roc_auc_score(y_test, y_proba_baseline),
        roc_auc_score(y_test, y_proba_tuned),
    ]
})

score_compares
```

### Refit (si besoin)

```python
final_model = search.best_estimator_

y_pred = final_model.predict(X_test)
y_proba = final_model.predict_proba(X_test)[:,1]
```

```python
display(search.best_params_,search.best_score_)
```

```python
scores_final = pd.DataFrame({
    "baseline": ["baseline_final"],
    "recall": [recall_score(y_test, y_pred)],
    "precision": [precision_score(y_test, y_pred)],
    "accuracy": [accuracy_score(y_test, y_pred)],
    "f1": [f1_score(y_test, y_pred)],
    "roc_auc": [roc_auc_score(y_test, y_proba)],
})


scores_final
```

## Question 1
### **Ton modèle prédit le churn. Pourquoi tu as choisi le recall comme métrique principale plutôt que l'accuracy ou la précision ?**

`Un faux négatif c'est un client qui va partir et qu'on n'a pas détecté — on perd ses 400€/an sans avoir rien tenté. 
Un faux positif c'est un client fidèle qu'on contacte inutilement — on lui envoie une promo, ça coûte 5€ et ça l'énerve peut-être un peu. 
Le coût n'est pas symétrique : rater un churner coûte 80x plus cher que se tromper sur un client fidèle. 
C'est pour ça que je maximise le recall — je préfère envoyer trop de promos que laisser partir des clients sans réagir.`

## Question 2
### **Tu as utilisé class_weight='balanced' dans ton meilleur modèle. C'est quoi l'effet concret sur l'entraînement ?**

`class_weight='balanced' augmente artificiellement le coût des erreurs sur la classe minoritaire pendant l'entraînement. 
C'est une façon de compenser le déséquilibre sans toucher aux données."`

## Question 3
### **Tu as fait un train_test_split puis un RandomizedSearchCV avec cv=5 sur le train. Pourquoi tu ne fais pas juste un split simple pour trouver les meilleurs paramètres ?**

`Un split simple pour choisir les hyperparamètres introduit du biais de sélection — je risque de choisir des paramètres qui marchent bien sur ce split précis mais qui généralisent mal. 
La CV me donne une estimation plus robuste de la vraie performance.`
